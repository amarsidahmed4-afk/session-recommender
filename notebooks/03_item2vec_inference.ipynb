```python
from gensim.models import Word2Vec
```

```python
# 1. Load the model from disk
model = Word2Vec.load('../data/models/item2vec.model')

# 2. Check the vocabulary size (how many unique products the model learned)
vocab = list(model.wv.key_to_index.keys())
print(f"Vocabulary size: {len(vocab)} unique products")

# 3. Grab one of the most popular products (e.g., the 10th most frequent item)
sample_product = vocab[10] 
print(f"\nFinding recommendations for product ID: {sample_product}")

# 4. Use the model to find the Top 5 most similar items (Nearest Neighbors)
similar_items = model.wv.most_similar(sample_product, topn=5)

for item, score in similar_items:
    print(f"Recommended Product: {item}, Similarity Score: {score:.4f}")
```
